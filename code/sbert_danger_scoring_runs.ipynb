# SBERT danger scoring of the training captions

Part of the code for *Danger Detection through Audio Captioning and Classification Fusion* (Electronics, MDPI). Paper sections: 3.3.2, 4.2 (Table 7).

Runs `sbert_danger_scoring.py` on the CoNeTTE captions of the 2058 training clips with the three threshold profiles (default, low, strict). The **low** profile (class 0.35, event 0.38, modifier 0.35) produced the labels used to fine-tune DistilBERT.

**Inputs.** Caption dataset (data/captions/), `sbert_lexicon/class_keywords.json`, `sbert_lexicon/danger_lexicon.json`. Sentence encoder: all-MiniLM-L6-v2 (sentence-transformers).

**Outputs.** Per-caption SBERT scores and labels, the DistilBERT-ready dataset, and per-class alignment summaries.

Paths point to a Google Drive layout; edit them to your own copy.
